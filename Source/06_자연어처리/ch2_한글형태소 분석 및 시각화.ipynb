## 3.1 HanNanum

```python
text = ''' 아름답지만 다소 복잡하기도 한 한국어는 전세계에서 13번째로 많이 사용되는 언어입니다.'''
```

```python
from konlpy.tag import Hannanum
hannanum = Hannanum(jvmpath=None,
                    max_heap_size=512) # 기본값: 1024M, 안쓰면 m(메가)로 인식됨. '1g' 최대 사이즈 4g(기가)
hannanum.analyze(text) # text를 세밀하게 분석해서, 형태소, 품사를 반환(ntag=69로 형태소 분석)
```

```python
# 형태소 분석: morphs
print(hannanum.morphs(text))
```

```python
# 명사만 추출 : nouns
print(hannanum.nouns(text))
```

```python
# 품사 태깅 : pos
print(hannanum.pos(text, ntags=22))
```

```python
# ex1. text에서 형용사만 추출
tagged_text = hannanum.pos(text, ntags=22)
# words = []
# for token, tag in tagged_text:
#     if tag == 'PA':
#         words.append(token)
words = [token for token, tag in tagged_text if tag=='PA']
print(words)
```

```python
# ex2. text에서 명사만 추출
tags = ['NC', 'NQ', 'NP', 'NN']
words = [token for token, tag in tagged_text if tag in tags]
words
```

```python
# ex3. text에서 보통명사(NC)만 추출
tags = ['NC']
words = [token for token, tag in tagged_text if tag in tags]
words
```

## 3.2 Kkma

```python
text = ''' 아름답지만 다소 복잡하기도 한 한국어는 전세계에서 13번째로 많이 사용되는 언어입니다.'''
```

```python
from konlpy.tag import Kkma
# Kkma = Kkma(jvmpath=None, max_heap_size='')
kkma = Kkma()
# 형태소 분석
print(kkma.morphs(text))
```

```python
# 품사 태깅 
print(kkma.pos(text)) # ntag 지정 불가 (무조건 56으로)
```

```python
# 품사태깅을 이용한 명사(NNG, NNP, NP)추출
tagged_text = kkma.pos(text)
words = [token for token, tag in tagged_text if tag in ('NNG', 'NNP', 'NP')]
print(words)
```

```python
# nouns함수를 이용한 명사 추출
print(kkma.nouns(text))
```
