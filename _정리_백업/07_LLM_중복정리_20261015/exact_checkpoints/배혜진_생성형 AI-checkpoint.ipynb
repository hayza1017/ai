# 5. 생성형 AI 평가
- 나라이름 → 그 나라에서 가장 유명한 음식(출력) → 그 음식의 레시피(출력)
1. 체인 = 음식 출력
2. 체인 = 음식의 레시피 출력

- 최종 체인
나라이름 → 그 나라에 가장 유명한 음식의 레시피

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
# llm 모델 생성
llm = ChatOllama(model='llama3.2:1b')
```

```python
# 음식 추천 invoke 구현
food_template = PromptTemplate(template = 'What is the famous food in {country}? Return the name of the food only.',
                    input_variables=['country'])
print('프롬프트 :', prompt)

output_parser = StrOutputParser()

first_chain = food_template | llm | output_parser
print('Parser 결과 :', first_chain.invoke({'country':'korea'}))
```

```python
# 음식 추천 invoke 구현
recipe_template = PromptTemplate(template = 'What is the famous recipe for {food}?',
                    input_variables=['food'])
print('프롬프트 :', prompt)

output_parser = StrOutputParser()

second_chain = recipe_template | llm | output_parser
print('Parser 결과 :', second_chain.invoke({'food':'Bibimbap'}))
```

```python
final_chain = first_chain | second_chain
```

```python
result = final_chain.invoke({'country':'Italy'})
```
